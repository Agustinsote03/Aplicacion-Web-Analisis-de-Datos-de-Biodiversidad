```python
import csv

with open(ruta_iadiza, "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f, delimiter="\t")
    count = 0
    for fila in reader:
        lat = fila.get("decimalLatitude", "").strip()
        if lat:
            print(f"lat: '{lat}'")
            count += 1
        if count == 10:
            break
    if count == 0:
        print("No hay ningún registro con latitud")
```
