```python
import sys
from pathlib import Path

#------IMPORTACIÓN DE FUNCIONES-------#

ruta_raiz = Path.cwd().parent
print(f"Ruta del directorio raiz: ", ruta_raiz)

ruta_de_src = Path(ruta_raiz / "src")
print(f"Ruta src: ", ruta_de_src)

# Se añade la ruta raíz al sys.path para poder importar módulos
sys.path.append(str(ruta_de_src))
from validador import validar_coordenadas, detectar_coordenadas_incompletas, validar_fechas, detectar_duplicados, detectar_incertidumbres_invalidas, detectar_country_codes_invalidos
#------IMPORTACIÓN DE DATASETS--------#

ruta_de_datasets = ruta_raiz / "raw_datasets"
print(f"Ruta de archivos: ", ruta_de_datasets)


ruta_iadiza = Path(ruta_de_datasets / "Colección Ornitológica del IADIZA-CCT CONICET Mendoza IADIZA-COI" / "occurrence.txt")
ruta_naturalist = Path(ruta_de_datasets / "inaturalist-filtered" / "observations.csv")
ruta_xeno_canto = Path(ruta_de_datasets / "xeno-canto-bird-sounds-filtered" / "Occurrence.txt")

#-------------CONFIGURACIÓN DE DATASETS------------#
CONFIGURACION_DATASETS = {
    "iadiza": {
        "ruta": ruta_iadiza,
        "separador": "\t",
        "col_lat": "decimalLatitude",
        "col_lon": "decimalLongitude"
    },
    "xeno_canto": {
        "ruta": ruta_xeno_canto,
        "separador": ",",
        "col_lat": "latitudeDecimal",
        "col_lon": "longitudeDecimal"
    },
    "inaturalist": {
        "ruta": ruta_naturalist,
        "separador": ",",
        "col_lat": "decimalLatitude",
        "col_lon": "decimalLongitude"
    }
}
```

Ejercicio 3.A- Información sobre registros inválidos



```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, registros = validar_coordenadas(cfg["ruta"], cfg["separador"], cfg["col_lat"], cfg["col_lon"])
    print(f"{nombre}: {cantidad} registros con coordenadas inválidas")
```

Ejercicio 3.B- Información sobre registros con coordenadas incompletas

```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, registros = detectar_coordenadas_incompletas(cfg["ruta"], cfg["separador"], cfg["col_lat"], cfg["col_lon"])
    print(f"{nombre}: {cantidad} registros con coordenadas incompletas")
```

Ejercicio 3.C - Validación y detección de fechas inválidas

```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, registros = validar_fechas(cfg["ruta"], "eventDate", cfg["separador"])
    print(f"{nombre}: {cantidad} registros con fechas inválidas")

#Nota: el formato de fecha esperado es "%d/%m/%y". Los datasets utilizan 
#formatos distintos (ej: "2025-05-03" o "2018-03-31T08:09:00-03:00"), 
#por lo que la mayoría de los registros son marcados como inválidos.
```

Ejercicio 3.D -  Información sobre registros duplicados

```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, ids = detectar_duplicados(cfg["ruta"], cfg["separador"])
    print(f"{nombre}: {cantidad} registros duplicados")
```

Ejercicio 3.E - Incertidumbres inválidas

```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, registros = detectar_incertidumbres_invalidas(cfg["ruta"], cfg["separador"])
    print(f"{nombre}: {cantidad} registros con incertidumbre inválida")
```

Ejercicio 3.F - Detección de CountryCodes inválidos

```python
for nombre, cfg in CONFIGURACION_DATASETS.items():
    cantidad, registros = detectar_country_codes_invalidos(cfg["ruta"], cfg["separador"])
    print(f"{nombre}: {cantidad} registros con country code inválido")

#Nota: Xeno-canto no utiliza el campo "countryCode" sino "country" con el 
#nombre completo del país (ej: "Argentina"). Por este motivo todos sus 
#registros son marcados como inválidos en esta validación.
```
