```python
import sys
from pathlib import Path

#----------------IMPORTACIONES----------------#

ruta_raiz = Path.cwd().parent
print(f"Ruta del directorio raiz: ", ruta_raiz)

ruta_src = Path(ruta_raiz / "src")
print(f"Ruta src: ", ruta_src)

#Agrego src al path
sys.path.append(str(ruta_src))

#Importo funciones
from dataset_reader import imprimir_primeras_filas, obtener_columnas, posiciones_columnas, cantidad_registros, columnas_con_nulos, porcentaje_nulos_por_columna, valores_distintos_columna, frecuencia_valores_columna, resumen_columna, columnas_totalmente_nulas

#------ DATASET -------#

ruta = ruta_raiz / "raw_datasets" / "IADIZA" / "occurrence.txt"
print(f"Ruta de archivos: ", ruta)

print("¿Existe el archivo?", ruta.exists())

#2A - Imprimo las primeras 10 filas del dataset
print ("Primeras 10 filas:")
imprimir_primeras_filas(ruta, "\t")

#2B - Imprimo la lista completa de columnas del dataset
print()
print ("Columnas:")
columnas = obtener_columnas(ruta)
print (columnas)

#2C - Imprimo nombre de la columna y su posición
print()
posiciones = posiciones_columnas(ruta)
print ("Posición de cada columna:")
print (posiciones)

#2D - Retorna la cantidad total de registros del dataset
print()
registros = cantidad_registros(ruta)
print("Cantidad de registros:", registros)
registros_esperados = 2370
if registros == registros_esperados:
    print("La cantidad de registros coincide con la documentación")
else:
    print("La cantidad de registros no coincide")
    print("Diferencia:", registros - registros_esperados)


#2E - Imprimo las columnas que poseen al menos un dato con registro nulo
print()
print("Columnas con al menos un dato nulo:")
columnas_nulas = columnas_con_nulos(ruta)
print(columnas_nulas)


#2F - Imprimo el porcentaje de registros con datos nulos para cada columna
print ()
print("Porcentaje de registros con datos nulos por columna:")
porcentajes_nulos = porcentaje_nulos_por_columna(ruta)
for columna, porcentaje in porcentajes_nulos.items():
    print(f"{columna}: {porcentaje:.2f}%")


#2G - Imprimo el número de valores distintos para la columna pasada por parámetro
print()
print (f"La cantidad de valores distintos en la columna son: {valores_distintos_columna(ruta, 'gbifID')}")


#2H - Imprimo la frecuencia de cada valor para la columna pasada por parámetro
print()
print("Frecuencia de cada valor en la columna:")
frecuencias =frecuencia_valores_columna(ruta, 'iucnRedListCategory')
for valor, cantidad in frecuencias.items():
    print (f"{valor}: {cantidad}")


#2I - Imprimo un resumen de la columna pasada por parámetro dependiendo de su tipo
print()
print ("Resumen de la columna (tipo texto):")
resultado = resumen_columna(ruta, "scientificName", "text")
for clave, valor in resultado.items():
    print (f"{clave}: {valor}")


#2J - Imprimo las columnas que tienen todos sus datos nulos
print()
columnas_nulas = columnas_totalmente_nulas(ruta)
print("Columnas completamente nulas:")
print(columnas_nulas)
```
