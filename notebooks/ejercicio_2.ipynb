Ejercicio 2.D - Retorna la cantidad total de registros del dataset y lo compara con la cantidad de registros informada en la documentación del dataset

```python
registros_esperados = {
    "iadiza": 2370,
    "xeno_canto": 13470,
    "inaturalist": 1229610
}

for nombre, cfg in CONFIGURACION_DATASETS.items():
    total = cantidad_registros(cfg["ruta"], cfg["separador"])
    print(f"\n{nombre}: {total} registros")
    esperados = registros_esperados[nombre]
    if esperados:
        if total == esperados:
            print("La cantidad coincide con la documentación")
        else:
            print(f"No coincide. Diferencia: {total - esperados}")
```

Ejercicio 2.F - Retorna para cada columna el porcentaje de registros nulos

Ejercicio 2.G - Dado el nombre de una columna, retorna la cantidad de valores distintos presentes en dicha columna

```python
columnas_a_consultar = {
    "iadiza": "gbifID",
    "xeno_canto": "id",
    "inaturalist": "id"
}

for nombre, cfg in CONFIGURACION_DATASETS.items():
    col = columnas_a_consultar[nombre]
    total = valores_distintos_columna(cfg["ruta"], col, cfg["separador"])
    print(f"\n{nombre} - Valores distintos en '{col}': {total}")
```

Ejercicio 2.H - Dado el nombre de una columna, retorna la frecuencia de aparición de cada valor

```python
columnas_a_consultar = {
    "iadiza": "iucnRedListCategory",
    "xeno_canto": "behavior",
    "inaturalist": "countryCode"
}

for nombre, cfg in CONFIGURACION_DATASETS.items():
    col = columnas_a_consultar[nombre]
    print(f"\n--- Frecuencia de '{col}' en {nombre} ---")
    frecuencias = frecuencia_valores_columna(cfg["ruta"], col, cfg["separador"])
    for valor, cantidad in frecuencias.items():
        print(f"{valor}: {cantidad}")
```

Ejercicio 2.I - Dada una columna y su tipo, retorne lo establecido para cada uno de los tipos

```python
columnas_a_consultar = {
    "iadiza": ("scientificName", "text"),
    "xeno_canto": ("scientificName", "text"),
    "inaturalist": ("scientificName", "text")
}

for nombre, cfg in CONFIGURACION_DATASETS.items():
    col, tipo = columnas_a_consultar[nombre]
    print(f"\n--- Resumen de '{col}' ({tipo}) en {nombre} ---")
    resultado = resumen_columna(cfg["ruta"], col, tipo, cfg["separador"])
    if resultado:
        for clave, valor in resultado.items():
            print(f"{clave}: {valor}")
```
