```python
import sys
from pathlib import Path

ruta_raiz = Path.cwd().parent

ruta_de_src = Path(ruta_raiz / "src")

sys.path.append(str(ruta_de_src))

from eliminacion_registros import eliminar_por_id_iadiza, eliminar_por_id_xeno_canto, eliminar_por_id_inaturalist, eliminar_por_lista_iadiza, eliminar_por_lista_xeno_canto, eliminar_por_lista_inaturalist, eliminar_por_condicion_iadiza, eliminar_por_condicion_xeno_canto, eliminar_por_condicion_inaturalist, sanitizar_dataset_iadiza, sanitizar_dataset_inaturalist, sanitizar_dataset_xeno_canto, sanitizar_dataset_evolucionada_iadiza, sanitizar_dataset_evolucionada_inaturalist, sanitizar_dataset_evolucionada_xeno_canto

# Importación de rutas de datos crudos
ruta_iadiza = Path(ruta_raiz / "raw_datasets" / "Colección Ornitológica del IADIZA-CCT CONICET Mendoza IADIZA-COI" / "occurrence.txt")
ruta_xeno = Path(ruta_raiz / "raw_datasets" / "xeno-canto-bird-sounds-filtered" / "Occurrence.txt")
ruta_inat = Path(ruta_raiz / "raw_datasets" / "inaturalist-filtered" / "observations.csv")

# Importación de rutas de datos procesados
ruta_iadiza_procesada = Path(ruta_raiz / "processed_datasets"/"occurrence_actualizado.txt")
ruta_procesada_xeno = Path(ruta_raiz / "processed_datasets" / "xeno_canto" / "Occurrence.txt")
ruta_procesada_inat = Path(ruta_raiz / "processed_datasets" / "inaturalist" / "observations.csv")

# Creo un diccionaro de filtros para buscar registros con countryCode igual a A
SEP_IADIZA = "\t"
SEP_XENO =  ","
SEP_INAT = ","

print(ruta_iadiza_procesada)
print(ruta_procesada_inat)
print(ruta_procesada_xeno)

nombre_dataset = "IADIZA"
```

Ejercicio 6.A - Eliminacion de registros segun su identificador.

```python
resultado_iadiza = eliminar_por_id_iadiza(ruta_iadiza, ruta_iadiza_procesada, "occurrenceID", "IADIZA:COI:006861")
print(f"IADIZA eliminado: {resultado_iadiza}")

resultado_xeno = eliminar_por_id_xeno_canto(ruta_xeno, ruta_procesada_xeno, "occurrenceID", "https://data.biodiversitydata.nl/xeno-canto/observation/XC999540")
print(f"Xeno-canto eliminado: {resultado_xeno}")

resultado_inat = eliminar_por_id_inaturalist(ruta_inat, ruta_procesada_inat, "occurrenceID", "https://www.inaturalist.org/observations/42369866")
print(f"iNaturalist eliminado: {resultado_inat}")
```

Ejercicio 6.B - Eliminacion de registros cuyo valor en una columna pertenezca a una lisa de valores suministrada por el usuario.

```python
resultado_iadiza = eliminar_por_lista_iadiza(ruta_iadiza, ruta_iadiza_procesada, "countryCode", ["AR", "CH"])
print(f"IADIZA eliminados: {resultado_iadiza}")

resultado_xeno = eliminar_por_lista_xeno_canto(ruta_xeno, ruta_procesada_xeno, "country", ["Argentina", "Chile"])
print(f"Xeno-canto eliminados: {resultado_xeno}")

resultado_inat = eliminar_por_lista_inaturalist(ruta_inat, ruta_procesada_inat, "countryCode", ["AR", "CL"])
print(f"iNaturalist eliminados: {resultado_inat}")
```

Ejercicio 6.C - Elimino registros si cumplen con las condiciones.

```python
resultado_iadiza = eliminar_por_condicion_iadiza(ruta_iadiza, ruta_iadiza_procesada, "stateProvince", "==", "Mendoza")
print(f"IADIZA eliminados: {resultado_iadiza}")

resultado_xeno = eliminar_por_condicion_xeno_canto(ruta_xeno, ruta_procesada_xeno, "country", "==", "Argentina")
print(f"Xeno-canto eliminados: {resultado_xeno}")

resultado_inat = eliminar_por_condicion_inaturalist(ruta_inat, ruta_procesada_inat, "stateProvince", "==", "Buenos Aires")
print(f"iNaturalist eliminados: {resultado_inat}")
```

Ejercicio 6.E - Sanitizar dataset.

```python
resumen_iadiza = sanitizar_dataset_iadiza(ruta_iadiza, ruta_iadiza_procesada)
print(f"IADIZA — total: {resumen_iadiza['total']}, eliminados: {resumen_iadiza['eliminados']}")

resumen_xeno = sanitizar_dataset_xeno_canto(ruta_xeno, ruta_procesada_xeno)
print(f"Xeno-canto — total: {resumen_xeno['total']}, eliminados: {resumen_xeno['eliminados']}")

resumen_inat = sanitizar_dataset_inaturalist(ruta_inat, ruta_procesada_inat)
print(f"iNaturalist — total: {resumen_inat['total']}, eliminados: {resumen_inat['eliminados']}")
```
